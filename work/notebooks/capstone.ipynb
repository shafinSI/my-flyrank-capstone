## 3. Methodology
Features used in the prototype include historical position, historical impressions, historical CTR, position volatility, URL depth, and in later prototype notebooks a smaller feature set including impression volume, position, and content age. The binary label is framed as “needs optimization / underperforming.”

## 6. Ranked recommendations
1. **Use the model only as a review queue.** Human editors should verify search intent, seasonality, and current SERP features before changing content.
2. **Prefer time-aware evaluation for search logs.** Report the final score on a future time window rather than relying only on random splits.
3. **Keep strict feature provenance.** Any feature that can be computed from post-outcome information should be excluded.
4. **Start with title/snippet review for high-CTR-gap candidates,** then consider content refresh/internal linking for rank-decay patterns.
5. **Monitor drift.** The Week 07 playbook proposes retraining when ROC-AUC drops below 0.80 and pausing after major search updates.
