# Week 02: ML Task Framing

## 1. My Lane as an ML Task (Type)
- **Task Type:** Binary Classification (CTR Opportunity Scoring).
- **Primary Goal:** Identify search pages with high impression volumes that receive significantly fewer clicks than expected for their position, so SEO teams can target them for title and meta description rewrites.

## 2. Target or Proxy
- **Target Variable ($y$):** `is_underperforming` (Binary Flag: `1` or `0`).
- **Logic:** A page is assigned `1` if its observed Click-Through Rate (CTR) falls below the position baseline curve despite high search impressions, and `0` otherwise.


## 3. Success Metric
- **Primary Metric:** `ROC-AUC` — Measures how effectively the model discriminates underperforming pages across all potential classification thresholds.
- **Secondary Metric:** `F1-Score` / `Precision@K` — Ensures high precision in the top recommendation list so SEO teams do not waste effort on false positives.

## 4. The Unit of Analysis, as a Real Dataframe
import duckdb
import pandas as pd

con = duckdb.connect()

query = """
SELECT 
    'https://example.com/blog/seo-guide' AS url,
    DATE '2026-08-01' AS date,
    1500 AS impressions,
    12 AS clicks,
    4.2 AS position,
    (12.0 / 1500.0) AS ctr,
    1 AS is_underperforming
UNION ALL
SELECT 
    'https://example.com/tools/keyword-planner' AS url,
    DATE '2026-08-01' AS date,
    3200 AS impressions,
    280 AS clicks,
    2.1 AS position,
    (280.0 / 3200.0) AS ctr,
    0 AS is_underperforming
"""

df_sample = con.execute(query).df()
print("Unit of Analysis: 1 Row = One Unique Page (URL) metrics on a specific Date")

## 5. Why ML Beats a Fixed Rule Here
- Search engine CTR decays exponentially relative to ranking position. A fixed rule (like `CTR < 2%`) fails because Position 1 and Position 9 have fundamentally different CTR expectations.
- Machine Learning models learn non-linear interactions between historical positions, impression volume, and expected CTR curves without relying on static thresholds that break during seasonality.
